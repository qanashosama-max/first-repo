```python
import cv2 
import pytesseract 
pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\tesseract-OCR\tesseract .exe'
import re
import tkinter as tk
from tkinter import filedialog

def extract_invoice_data(image_path):
    img = cv2.imread(image_path)
    if img is None:
        return {"Error": "لا يمكن قراءة الصورة. تأكد من الملف"}
        
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    text = pytesseract.image_to_string(gray, lang='ara+eng')
    lines = [line.strip() for line in text.split('\n') if line.strip()]
    
    data = {
        "Company": lines[0] if lines else "",
        "Date": "",
        "Total": "",
        "Items": []
    }   
      
    date_pattern = r'\b\d{1,4}[-/]\d{1,2}[-/]\d{1,4}\b'
    total_pattern = r'(total|المجموع|الإجمالي)[\s:\-]*([\d\.,]+)'

    for line in lines:
        if not data["Date"]:
            d_match = re.search(date_pattern, line)
            if d_match:
                data["Date"] = d_match.group()
                
        t_match = re.search(total_pattern, line, re.IGNORECASE)
        if t_match and not data["Total"]:
            data["Total"] = t_match.group(2)
            continue
            
        if re.search(r'\d', line) and len(line) > 5 and not t_match:
            data["Items"].append(line)
            
    return data

def select_image():
    root = tk.Tk()
    root.withdraw() 
    root.attributes('-topmost', True)
    file_path = filedialog.askopenfilename(
        title="اختر صورة الفاتورة",
        filetypes=[("Image files", "*.jpg *.jpeg *.png")]
    )
    root.destroy()
    return file_path

if __name__ == "__main__":
    print("الرجاء اختيار صورة الفاتورة من النافذة المنبثقة...")
    image_file = select_image()

    if image_file:
        invoice_info = extract_invoice_data(image_file)
        print("\n--- البيانات المستخرجة ---")
        for key, value in invoice_info.items() :
            if isinstance(value, list):
                print(f"{key}:")
                for item in value:
                    print(f" - {item}") 
            else:
                print(f"{key}: {value}")
    else:
       print("لم يتم اختيار أي صورة، تم إنهاء البرنامج.") 
```
